Notebook ini membahas pencarian akar untuk persamaan nonlinear **satu variabel**. Jalankan cell dari atas ke bawah.

## 1. Dasar teori dan contoh

Kita mencari akar dari:

$$f(x) = x^3 + 2x - 7 = 0$$

Interval awal yang dipakai adalah `[1, 2]`. Nilai ujung interval adalah `f(1) = -4` dan `f(2) = 5`. Karena tanda kedua nilai berbeda dan polinomial kontinu, interval ini mengurung setidaknya satu akar. Bisection berulang kali membagi interval menjadi dua dan mempertahankan bagian yang memiliki perubahan tanda.

Tiga titik tengah pertama sebagai ilustrasi: `c₁=1,5` dengan `f(c₁)=−0,625`, `c₂=1,75` dengan `f(c₂)=1,859375`, lalu `c₃=1,625` dengan `f(c₃)=0,541015625`. Setelah masing-masing evaluasi, interval dipersempit pada bagian yang mempertahankan perubahan tanda.

## 2. Menentukan fungsi dan parameter

Cell ini menetapkan fungsi `f`, batas kiri `a`, batas kanan `b`, toleransi, serta jumlah iterasi maksimum. Toleransi mengendalikan ketelitian penghentian; iterasi maksimum mencegah loop berjalan tanpa batas.

```python
# Definisikan fungsi nonlinear yang akarnya akan dicari.
def f(x):
    return x**3 + 2*x - 7

# Tetapkan interval awal yang diharapkan mengurung akar.
a_awal = 1.0
b_awal = 2.0

# Tetapkan toleransi untuk nilai fungsi atau batas galat interval.
toleransi = 1e-6

# Batasi jumlah iterasi agar program selalu memiliki batas kerja.
iterasi_maksimum = 100

# Perlihatkan nilai fungsi pada kedua ujung interval sebelum iterasi.
print(f"f(a) = f({a_awal}) = {f(a_awal):.6f}")
print(f"f(b) = f({b_awal}) = {f(b_awal):.6f}")
```

## 3. Implementasi fungsi Bisection

Fungsi berikut menjalankan prosedur pembagian interval. Program terlebih dahulu memvalidasi bahwa batas bawah lebih kecil daripada batas atas dan hasil kali nilai fungsi di kedua ujung tidak positif. Pada setiap iterasi, program menghitung titik tengah, nilai fungsi di titik tengah, serta setengah lebar interval sebagai batas galat posisi akar. Riwayat iterasi disimpan agar dapat ditampilkan sebagai tabel.

```python
def metode_bisection(fungsi, a, b, toleransi=1e-6, iterasi_maksimum=100):
    # Pastikan interval diberikan dengan urutan batas yang benar.
    if a >= b:
        raise ValueError("Batas bawah a harus lebih kecil daripada batas atas b.")

    # Hitung nilai fungsi pada kedua ujung interval.
    fa = fungsi(a)
    fb = fungsi(b)

    # Pastikan akar diapit oleh interval berdasarkan perubahan tanda.
    if fa * fb > 0:
        raise ValueError("f(a) dan f(b) harus memiliki tanda berlawanan.")

    # Buat list untuk menyimpan data setiap iterasi.
    riwayat = []

    # Jika salah satu ujung sudah menjadi akar dalam toleransi, kembalikan langsung.
    if abs(fa) <= toleransi:
        return a, riwayat, True
    if abs(fb) <= toleransi:
        return b, riwayat, True

    # Jalankan pembagian interval hingga toleransi tercapai atau iterasi habis.
    for iterasi in range(1, iterasi_maksimum + 1):
        # Hitung titik tengah interval sebagai perkiraan akar saat ini.
        c = (a + b) / 2.0

        # Evaluasi fungsi pada titik tengah tersebut.
        fc = fungsi(c)

        # Setengah panjang interval menjadi batas galat posisi akar.
        batas_galat = (b - a) / 2.0

        # Catat seluruh informasi iterasi sebelum interval diperbarui.
        riwayat.append({
            "iterasi": iterasi,
            "a": a,
            "b": b,
            "c": c,
            "f(c)": fc,
            "batas_galat": batas_galat,
        })

        # Hentikan jika nilai fungsi atau batas galat sudah memenuhi toleransi.
        if abs(fc) <= toleransi or batas_galat <= toleransi:
            return c, riwayat, True

        # Jika perubahan tanda ada di [a, c], pertahankan interval bagian kiri.
        if fa * fc < 0:
            b = c
            fb = fc

        # Jika tidak, perubahan tanda ada di [c, b], maka pertahankan bagian kanan.
        else:
            a = c
            fa = fc

    # Jika loop selesai, kembalikan titik tengah terakhir dan status belum konvergen.
    return c, riwayat, False
```

## 4. Menjalankan Bisection

Panggil fungsi dengan fungsi nonlinear, interval awal, toleransi, dan batas iterasi. Variabel `akar` menyimpan perkiraan akar, `data_iterasi` menyimpan tabel proses, dan `konvergen` menunjukkan apakah kriteria berhenti terpenuhi.

```python
# Jalankan metode Bisection menggunakan parameter yang telah ditentukan.
akar, data_iterasi, konvergen = metode_bisection(
    f, a_awal, b_awal, toleransi, iterasi_maksimum
)

# Beri tahu apakah program berhenti karena toleransi tercapai.
print("Status konvergensi:", "Tercapai" if konvergen else "Belum tercapai")
print(f"Jumlah iterasi: {len(data_iterasi)}")
print(f"Perkiraan akar: {akar:.10f}")
print(f"Nilai fungsi pada akar pendekatan: {f(akar):.12e}")
```

## 5. Menampilkan tabel iterasi

Kolom `a` dan `b` menunjukkan interval pada awal iterasi, `c` adalah titik tengah, `f(c)` adalah nilai fungsi pada titik tengah, dan `batas_galat` adalah setengah lebar interval. Tabel ini menjadi bukti bahwa proses penyempitan interval dilakukan secara bertahap.

```python
# Cetak judul kolom agar data iterasi mudah dibaca.
print(f"{'Iterasi':>7} {'a':>12} {'b':>12} {'c':>12} {'f(c)':>14} {'Batas galat':>14}")

# Tampilkan setiap catatan iterasi dengan format angka yang konsisten.
for baris in data_iterasi:
    print(
        f"{baris['iterasi']:7d} "
        f"{baris['a']:12.8f} "
        f"{baris['b']:12.8f} "
        f"{baris['c']:12.8f} "
        f"{baris['f(c)']:14.6e} "
        f"{baris['batas_galat']:14.6e}"
    )
```

## 6. Pemeriksaan hasil dan interpretasi

Substitusikan akar pendekatan ke fungsi untuk melihat apakah nilainya mendekati nol. Nilai tersebut bukan harus tepat nol karena metode numerik menghasilkan pendekatan yang ketelitiannya ditentukan oleh toleransi.

```python
# Hitung ulang nilai fungsi pada akar pendekatan sebagai verifikasi.
nilai_verifikasi = f(akar)

# Ambil batas galat interval dari catatan iterasi terakhir.
batas_galat_akhir = data_iterasi[-1]["batas_galat"] if data_iterasi else 0.0

# Tampilkan nilai akar, residual fungsi, dan batas galat interval secara terpisah.
print(f"Akar pendekatan       = {akar:.10f}")
print(f"f(akar)               = {nilai_verifikasi:.12e}")
print(f"Batas galat interval  = {batas_galat_akhir:.12e}")
print(f"Toleransi              = {toleransi:.1e}")

# Beri kesimpulan sederhana berdasarkan kriteria konvergensi yang dikembalikan.
if konvergen:
    print("Kesimpulan: nilai fungsi atau batas galat interval telah memenuhi toleransi.")
else:
    print("Kesimpulan: iterasi maksimum tercapai sebelum toleransi terpenuhi.")
```

Metode Bisection mempersempit interval yang mengapit akar dengan membagi interval menjadi dua pada setiap iterasi. Untuk contoh ini, akar berada di sekitar `1,568946` (angka akhir bergantung pada toleransi).
